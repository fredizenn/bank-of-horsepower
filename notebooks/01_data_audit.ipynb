### Q2. `season` is unique in `f1_business`
`groupby().nunique()` (cell above) shows 1 per column even if a season appears twice with identical values, so count rows per season directly.

```python
rows_per_season = f_one_business_df["season"].value_counts()

print("# of rows: ", len(f_one_business_df))
print("# of unique seasons: ", f_one_business_df["season"].nunique())
print("Seasons with more than one row: ", rows_per_season[rows_per_season > 1].index.tolist())
```

### Q4. `f1_business` revenue components are a fixed split of the total

```python
component_columns = ["race_promotion_fees_usd_m", "broadcast_media_usd_m", "sponsorship_usd_m", "other_revenue_usd_m"]

component_shares = f_one_business_df.set_index("season")[component_columns].div(f_one_business_df.set_index("season")["total_revenue_usd_m"], axis=0).round(3)
component_shares
```

### Q5. Does `prize_money_usd_m` follow this season's or last season's champion?
Rank of the champion's prize money within each season. Neither column settles it (see Q5 open question).

```python
prize_check = constructor_df.sort_values(["team_name", "season"]).copy()
prize_check["was_champion_last_season"] = prize_check.groupby("team_name")["is_constructors_champion"].shift(1)
prize_check["prize_money_rank"] = prize_check.groupby("season")["prize_money_usd_m"].rank(ascending=False)

print("Champion this season, prize money rank:")
print(prize_check.loc[prize_check["is_constructors_champion"] == 1, "prize_money_rank"].value_counts())
print("Champion last season, prize money rank:")
print(prize_check.loc[prize_check["was_champion_last_season"] == 1, "prize_money_rank"].value_counts())
```
